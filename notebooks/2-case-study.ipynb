# Case Study: Midair Collision Over the Hudson River

Reading about safety frameworks is important, but it is easy to forget why they matter until you
see what happens when safety procedures break down. This learning experience walks through a real
midair collision to make the risks of operating in shared airspace concrete.

## Intended learning outcomes

After completing this learning experience, learners will be able to:

1. Describe the chain of human and procedural failures that contributed to a real aviation accident.
2. Explain why the "see-and-avoid" concept has inherent limitations for pilots sharing airspace.
3. Connect lapses in air traffic control procedure to accident risk.

## Background

On August 8, 2009, a Piper PA-32R-300 airplane and a Eurocopter AS350BA sightseeing helicopter
collided over the Hudson River near Hoboken, New Jersey. All 9 people aboard the two aircraft were
killed. The National Transportation Safety Board (NTSB) investigated the accident and identified
several contributing factors, including the inherent limitations of the see-and-avoid concept and
a lapse in air traffic control procedure.
