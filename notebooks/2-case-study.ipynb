# Case Study: Midair Collision Over the Hudson River

Reading about safety frameworks is important, but it is easy to forget why they matter until you
see what happens when safety procedures break down. This learning experience walks through a real
midair collision to make the risks of operating in shared airspace concrete.

## Intended learning outcomes

After completing this learning experience, learners will be able to:

1. Describe the chain of human and procedural failures that contributed to a real aviation accident.
2. Explain why the "see-and-avoid" concept has inherent limitations for pilots sharing airspace.
3. Connect lapses in air traffic control procedure to accident risk.

## Background

On August 8, 2009, a Piper PA-32R-300 airplane and a Eurocopter AS350BA sightseeing helicopter
collided over the Hudson River near Hoboken, New Jersey. All 9 people aboard the two aircraft were
killed. The National Transportation Safety Board (NTSB) investigated the accident and identified
several contributing factors, including the inherent limitations of the see-and-avoid concept and
a lapse in air traffic control procedure.

## Self-Assessment

Run the cell below once to load the checker, then run each question's cell and select your
answer(s). You'll see immediately whether you got it right — if not, try again.

```python
import ipywidgets as widgets
from IPython.display import display


def make_multi_select_question(prompt, options, correct_indices):
    checkboxes = [widgets.Checkbox(value=False, description=opt, indent=False,
                                    layout=widgets.Layout(width="100%")) for opt in options]
    button = widgets.Button(description="Check answer", button_style="primary")
    output = widgets.Output()

    def on_click(_):
        output.clear_output()
        selected = {i for i, cb in enumerate(checkboxes) if cb.value}
        with output:
            if selected == set(correct_indices):
                print("\u2705 Correct!")
            else:
                print("\u274c Not quite — uncheck/check some boxes and try again.")

    button.on_click(on_click)
    display(widgets.HTML(f"<b>{prompt}</b>"), *checkboxes, button, output)


def make_true_false_question(prompt, correct_answer):
    toggle = widgets.ToggleButtons(options=["True", "False"], style={"button_width": "100px"})
    button = widgets.Button(description="Check answer", button_style="primary")
    output = widgets.Output()

    def on_click(_):
        output.clear_output()
        with output:
            if (toggle.value == "True") == correct_answer:
                print("\u2705 Correct!")
            else:
                print("\u274c Not quite — try again.")

    button.on_click(on_click)
    display(widgets.HTML(f"<b>{prompt}</b>"), toggle, button, output)
```

### Question 1

What factors probably contributed to the accident? (Select all that apply.)

```python
make_multi_select_question(
    "Select all factors that probably contributed to the accident:",
    [
        "The helicopter pilot was flying too high.",
        "The airplane pilot did not see the helicopter.",
        "The airplane pilot was too tired.",
        "The airport controller was having a non-pertinent phone call.",
    ],
    correct_indices={1, 3},
)
```

### Question 2

True or false? If the helicopter pilot had been flying lower, the accident probably would not
have happened.

```python
make_true_false_question(
    "If the helicopter pilot had been flying lower, the accident probably would not have happened.",
    correct_answer=False,
)
```

### Question 3

True or false? If the airport's manager had talked to the controller about the earlier
non-pertinent phone call, the accident probably would not have happened.

```python
make_true_false_question(
    "If the airport's manager had talked to the controller about the earlier non-pertinent phone "
    "call, the accident probably would not have happened.",
    correct_answer=True,
)
```
