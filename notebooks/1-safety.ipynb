# Safety

Before you touch a drone, a soldering iron, or fly outdoors, you need to understand the risks
involved and how to manage them. This learning experience walks you through the safety analysis
you must complete before starting the *Introduction to Robotics with Drones* course.

## Intended learning outcomes

After completing this learning experience, learners will be able to:

1. Analyze the safety considerations for operating a robot (drone) using the OSHA framework for
   industrial robot and robot system safety.
2. Explain the FAA rules and regulations that govern operating an unmanned aircraft system (UAS),
   including how to complete the TRUST certification.
3. Identify the institutional (Brown University) and personal (at-home) policies and precautions
   that apply to flying a drone outside of the lab.
4. Set up a soldering station and a flight area that meet the safety requirements for this course.



## 1. OSHA Safety Analysis

Robots — including drones — can injure the people operating and working around them if they are
not used correctly. The Occupational Safety and Health Administration (OSHA) publishes a Technical
Manual on Industrial Robots and Robot System Safety that lays out the standard framework for
thinking about robot safety.

**Read:** the OSHA Technical Manual, Section IV, Chapter 4 — *Industrial Robots and Robot System
Safety* (search for it on osha.gov if the link your instructor gave you has moved).

**Do:** Prepare a slide deck discussing how the OSHA framework to the drone you will fly in this course, answer the following.

1. What tasks will the robot perform?
2. What are the startup, command, and programming procedures?
3. What are the relevant environmental conditions?
4. What are the location / installation requirements?
5. What potential human errors could occur?
6. What maintenance is required?
7. What robot and system malfunctions are possible?
8. What is the normal operation mode?

Write specific, concrete answers about *your* drone and *your* lab space — generic answers copied
from the OSHA manual will not earn full credit.


## 2. FAA Rules

Because a drone is an unmanned aircraft, flying one is regulated by the Federal Aviation
Administration (FAA), not just by OSHA-style workplace safety rules.

**Do:**

1. Review the FAA's [Unmanned Aircraft Systems](https://www.faa.gov/uas) website.
2. Complete **The Recreational UAS Safety Test (TRUST)** through any FAA-approved test
   administrator. This is free and takes about 30 minutes. Keep the certificate and show it as part of your checkoff.
3. Answer the following four questions:
   1. What are the procedures for flying outdoors, beyond the CIT (or wherever your indoor lab
      space is)?
   2. What is the nearest airport to campus, including heliports?
   3. What are the risks and potential injuries involved in flying a drone?
   4. What are the FAA's requirements for reporting an accident?

Your photo must clearly show:

- A window or ventilation fan (to clear solder fumes)
- A fire extinguisher nearby
- A surge-protected power source
